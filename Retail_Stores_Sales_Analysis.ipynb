```python
df["Sales"] = (df["Quantity"] * df["Unit_Price"]) - df["Discount"]

print(df.head())
```

```python
#Question1
total_sales = np.sum(df["Sales"])

print("Total Sales =", total_sales)
```

```python
df["Month"] = df["Date"].dt.month_name()

monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
#Question2
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales.sort_values(ascending=False))
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

print(product_sales.sort_values(ascending=False))

print("Top Product =", product_sales.idxmax())

print("Lowest Product =", product_sales.idxmin())
```

```python
#Question3
branch_sales = df.groupby("Branch")["Sales"].sum()

print(branch_sales)

print("Best Branch =", branch_sales.idxmax())

print("Worst Branch =", branch_sales.idxmin())
```

```python
#Question4
df["Weekday"] = df["Date"].dt.day_name()

weekday_sales = df.groupby("Weekday")["Sales"].sum()

print(weekday_sales.sort_values(ascending=False))
```

```python
#Question5
avg_transaction = np.mean(df["Sales"])

print("Average Transaction Value =", avg_transaction)
```

```python
#Question6
correlation = df["Discount"].corr(df["Sales"])

print("Discount and Sales Correlation =", correlation)
```

```python
#Question7
customer_sales = df.groupby("Customer_ID")["Sales"].sum()

print(customer_sales.sort_values(ascending=False).head(10))
```
